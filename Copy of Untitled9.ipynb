```python
import random

# Snakes and their positions
snakes = {
    99: 54,
    87: 36,
    64: 25,
    48: 10,
    32: 8
}

# Ladders and their positions
ladders = {
    4: 25,
    13: 46,
    27: 56,
    42: 75,
    58: 89
}


# Function to roll the dice
def roll_dice():
    return random.randint(1, 6)


# Function to move the player
def move_player(position, dice):
    new_position = position + dice

    if new_position > 100:
        print("You cannot move!")
        print("You need an exact number to reach 100.")
        return position

    position = new_position

    # Check for ladder
    if position in ladders:
        print("You found a ladder!")
        position = ladders[position]
        print("You climbed to:", position)

    # Check for snake
    elif position in snakes:
        print("Oh no! You found a snake!")
        position = snakes[position]
        print("You came down to:", position)

    return position


# Function to display positions
def display_positions(name1, name2, pos1, pos2):
    print("-------------------------------")
    print(name1, ":", pos1)
    print(name2, ":", pos2)
    print("-------------------------------")


# Function to play the game
def play_game():

    print("\n================================")
    print("       SNAKE AND LADDER")
    print("================================")

    name1 = input("Enter Player 1 name: ")
    name2 = input("Enter Player 2 name: ")

    position1 = 0
    position2 = 0

    turn = 1

    print("\nGame starts!")
    print("First player to reach 100 wins.")
    print()

    while True:

        # Player 1 turn
        if turn == 1:

            print(name1, "'s turn")

            input("Press Enter to roll the dice...")

            dice = roll_dice()

            print("You rolled:", dice)

            position1 = move_player(position1, dice)

            print(name1, "position:", position1)

            if position1 == 100:
                print("\n************************")
                print(name1, "WINS!")
                print("************************")
                break

            turn = 2

        # Player 2 turn
        else:

            print(name2, "'s turn")

            input("Press Enter to roll the dice...")

            dice = roll_dice()

            print("You rolled:", dice)

            position2 = move_player(position2, dice)

            print(name2, "position:", position2)

            if position2 == 100:
                print("\n************************")
                print(name2, "WINS!")
                print("************************")
                break

            turn = 1

        display_positions(
            name1,
            name2,
            position1,
            position2
        )


# Start the game
while True:

    play_game()

    print("\nDo you want to play again?")

    choice = input("Enter yes or no: ")

    if choice.lower() == "no":
        print("Thank you for playing!")
        break

    elif choice.lower() == "yes":
        print("\nStarting a new game...")

    else:
        print("Invalid choice!")
        print("Game will end.")
        break
```
